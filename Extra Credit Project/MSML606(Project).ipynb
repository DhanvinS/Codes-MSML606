Using a dataset with around 25k samples (Subreddits)

```python
# Cleaning the data

# keeping only required columns
df = df[['name', 'n_members']]

# dropping missing names
df = df.dropna(subset=['name'])

# converting everything to lowercase
df['name'] = df['name'].str.lower()

# removing duplicates
df = df.drop_duplicates(subset=['name'])

# filling up missing members
df['n_members'] = df['n_members'].fillna('0')

print("Cleaned rows:", len(df))
df.head()
```

```python
class TreeNode:
    def __init__(self, name, members):
        self.name = name
        self.members = members
        self.left = None
        self.right = None

```

```python
class BST:
    def __init__(self):
        self.root = None

    def insert(self, root, name, members):
        if root is None:
            return TreeNode(name, members)

        if name < root.name:
            root.left = self.insert(root.left, name, members)
        elif name > root.name:
            root.right = self.insert(root.right, name, members)

        return root

    def search(self, root, name):
        if root is None or root.name == name:
            return root

        if name < root.name:
            return self.search(root.left, name)
        else:
            return self.search(root.right, name)

    def inorder(self, root, result):
        if root:
            self.inorder(root.left, result)
            result.append((root.name, root.members))
            self.inorder(root.right, result)
```

```python
bst = BST()

for _, row in df.iterrows():
    bst.root = bst.insert(
        bst.root,
        row['name'],
        row['n_members']
    )

print("BST has been built")
```

```python
search_name = input("Enter subreddit name: ").lower()

result = bst.search(bst.root, search_name)

if result:
    print("\nFound!")
    print("Name:", result.name)
    print("Members:", result.members)
else:
    print("Not found")
```

```python
sorted_list = []
bst.inorder(bst.root, sorted_list)

print("\nFirst 20 sorted subreddits:")
for item in sorted_list[:20]:
    print(item)
```

```python
import time

start = time.time()
bst.search(bst.root, search_name)
end = time.time()

print("Search time:", end - start, "seconds")
```
